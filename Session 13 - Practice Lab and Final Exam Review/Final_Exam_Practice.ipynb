# Section D — Supervised learning (20 points)

A car maker studies older models before designing new ones. `cars` has one row per car model; `weight` is in
pounds and `mpg` is miles per gallon. `high_mileage` is True when `mpg` is 30 or more. `origin` takes three
values: `usa`, `japan` and `europe`.

The next cell builds `cars` from seaborn's `mpg` data; you do not need to read it.

```python
cars = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv")
cars = cars[['name', 'model_year', 'origin', 'cylinders', 'weight', 'mpg']]
cars['high_mileage'] = cars['mpg'] >= 30
```

```python
cars.head(3)
```

## Question D1 (10 points)

**(a)** (4 points) Which statements are true? (choose all that apply)

- A. A car with a missing `origin` gets `True` in one of the `origin_` columns, chosen by default.
- B. If `weight` were also listed in `columns=`, `get_dummies` would group the weights into a few bins.
- C. Running `X = pd.get_dummies(...)` leaves `cars` unchanged, still with its one `origin` column.
- D. If `origin` had a fourth value, `X` would have four `origin_` columns.

```python
X = pd.get_dummies(cars[['weight', 'origin']], columns=['origin'])
X.head(3)
```

```python
# Check it: one car from each origin, and how many origin_ columns are True per row
print(X[cars.origin == 'europe'].head(1))
print(X[cars.origin == 'japan'].head(1))
print(cars.columns.tolist())

# A car with a missing origin: every origin_ column is False
print(pd.get_dummies(pd.DataFrame({'origin': ['usa', None]}), columns=['origin']))
```

**(b)** (3 points) Which statements about this code are true? (choose all that apply)

- A. `max_depth=3` lets the tree ask at most three questions before it reaches a prediction.
- B. The tree is built from 80% of the cars.
- C. `random_state=595` makes this tree more accurate than other values would.
- D. `model.fit(X_train, y_train)` also looks at the test cars when it builds its rules.

**(c)** (3 points) The company will advertise every car that the model predicts to be high-mileage. It wants as few of those adverts as possible to be for cars that are not really high-mileage. Which number best shows how well the model does at that? (choose one)

- A. 0.887
- B. 14 / 19
- C. 14 / 18
- D. 57 / 62

```python
X = pd.get_dummies(cars[['weight', 'model_year', 'cylinders', 'origin']],
                   columns=['origin'])
y = cars['high_mileage']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=595, stratify=y)

model = DecisionTreeClassifier(max_depth=3, min_samples_leaf=20, random_state=595)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("accuracy:", round(accuracy_score(y_test, y_pred), 3))
print(pd.DataFrame(confusion_matrix(y_test, y_pred),
                   index=["actual False", "actual True"],
                   columns=["predicted False", "predicted True"]))
```

```python
# Check it: precision and recall for the high-mileage class
from sklearn.metrics import precision_score, recall_score
print(round(precision_score(y_test, y_pred), 3), round(recall_score(y_test, y_pred), 3))
```

```python
# Check it: the tree's depth, and how many cars fit used versus kept back
from sklearn.tree import plot_tree
print(model.get_depth(), len(X_train), len(X_test))
plot_tree(model, feature_names=list(X_train.columns), filled=True)
plt.show()
```

## Question D2 (10 points)

The analyst now predicts `mpg` itself.

**(a)** (4 points) Which statements are true? (choose all that apply)

- A. The tree with `max_depth=1` is underfitting.
- B. The tree with `max_depth=16` is overfitting.
- C. `max_depth=16` is the best choice, because it fits the training cars perfectly.
- D. `max_depth=8` is a better choice than `max_depth=4`.

```python
X = cars[['weight', 'model_year']]
y = cars['mpg']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=595)

rows = []
for depth in [1, 2, 4, 8, 16]:
    tree = DecisionTreeRegressor(max_depth=depth, random_state=595)
    tree.fit(X_train, y_train)
    rows.append({"max_depth": depth,
                 "train R2": r2_score(y_train, tree.predict(X_train)),
                 "test R2": r2_score(y_test, tree.predict(X_test))})
pd.DataFrame(rows).round(2)
```

**(b)** (3 points) What will the histogram show? (choose one)

- A. A single bar at 0: the two models give every test car the same prediction.
- B. A spread of small differences around 0, because each fit starts from random values.
- C. Two separate bars, one for each model's predictions.
- D. No chart: the code stops with an error, because the same data cannot be fit twice.

Answer on paper first, then run the cell.

```python
model_a = LinearRegression()
model_a.fit(X_train, y_train)
model_b = LinearRegression()
model_b.fit(X_train, y_train)
diff = model_a.predict(X_test) - model_b.predict(X_test)
sns.histplot(diff)
plt.show()
```
