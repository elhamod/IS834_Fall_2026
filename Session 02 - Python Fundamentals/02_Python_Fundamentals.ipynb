# Strings

- `'brock'` `==` `"brock"` are both strings --> Note the use of single or double quotes, either are fine, but we can't mix
- python __is case sensitive!__  `brock` != `Brock` != `BROCK`

__Notes:__
- strings have a length
- strings are sequences so we can `slice` into them (i.e. extract contents by indexing)
- strings have methods to inspect, change case, etc.


```python
# A string is text wrapped in quotes. Single or double both work, as long as you do not mix them in one string.
myname = "Brock"
print(myname)
```

```python
# len() counts the characters in a string.
len(myname)
```

> `len` is common across various python data types!


```python
# .lower() and .upper() hand back a NEW string -- the original variable is untouched.
print(myname.lower())
print(myname.upper())
print(myname)
```

```python
# Store the lowercased version if you want to keep it around.
myname_lower = myname.lower()
myname_lower
```

```python
# Python is case sensitive, so the two spellings are not the same value.
myname == myname_lower
```

```python
# + pastes strings together -- notice that you have to supply the space yourself.
'Brock' + ' ' + 'likes to code in Python'
```

```python
# Two string literals written side by side are joined automatically.
# This only works for literals typed out in your code, never for variables.
'Brock'  ' '  'likes to code in Python'
```

```python
# .count() reports how many times a sequence of characters shows up.
greeting = "Hello IS834"
print(greeting.count("l"))
print(greeting.count("e"))
```

```python
# .startswith() and .endswith() ask a yes-or-no question about the ends of a string.
print(greeting.endswith("834"))
print(greeting.startswith("Hello"))
```

```python
# Indexing into a string works by position, and position 0 is the first character.
mystring = "IS 834"

print(mystring[0])      # a single character
print(mystring[0:2])    # positions 0 and 1 -- the end position is NOT included
print(mystring[:2])     # leaving the start out means "from the beginning"
print(mystring[3:5])
```

```python
# .split() with no argument breaks the text apart on spaces and returns a list.
mystring.split()
```

```python
# Pass a character to .split() when the data uses something other than a space to separate the pieces.
newval = "Brock-likes-to-code-in Python"
newval.split("-")
```

```python
# .replace() swaps every occurrence of one piece of text for another and returns a new string.
address = "123 Main Street"

print(address.replace("Street", "St."))
print(address)   # unchanged -- string methods never edit the original in place
```

## Lists
![](https://pynative.com/wp-content/uploads/2021/03/python-list.jpg)


```python
# A list is written with square brackets -- a useful visual tip when reading someone else's code.
fruits = ['apple', 'cherry', 'grapes']
fruits
```

```python
# len() works on a list too, and counts items rather than characters.
len(fruits)
```

```python
# Slicing a list follows the same start:end rule as a string.
fruits[:2]
```

```python
# .append() adds one item to the end of the list.
fruits.append("pear")
```

<div style="
    background-color: green;

    padding: 10px;

">
    <strong>Note:</strong> This is the first time we have seen an "in-place" operation.  We didn't assign the list, we modified the __object__ fruits in place.

```python
# The list changed even though we never reassigned it.
fruits
```

```python
# Assigning to a position replaces whatever was there.
fruits[1] = 'banana'
fruits
```

```python
# A list can hold any Python object, including another list. Very powerful, and occasionally confusing.
numbers = [1, 2, 3]
fruits.append(numbers)

fruits
```

```python
# The nested list counts as ONE item, not three.
len(fruits)
```

```python
# + joins two lists and builds a brand new one; the originals are untouched.
list1 = list(range(3))     # range(3) gives 0, 1, 2
list2 = list("brock")      # a string turns into a list of its characters

print(list1)
print(list2)
print(list1 + list2)
```

```python
# .extend() does the same join, but modifies list1 in place instead of returning a new list.
list1.extend(list2)
print(list1)
```

![](https://pynative.com/wp-content/uploads/2021/02/dictionaries-in-python.jpg)


```python
# A dictionary uses curly braces and stores key: value pairs -- you look things up by name, not by position.
student = {"name": "Brock", "age": 21, "fav_sports": ["hockey", "golf"]}

print(student['name'])
print(student['fav_sports'])
```

# For Loops

![](https://www.learnbyexample.org/wp-content/uploads/python/Python-for-Loop-Syntax.png)


```python
# A for loop runs the indented body once for each item in a collection.
x = list(range(5))

for i in x:
    print(i)
```

```python
# Naming the loop variable after one item of the collection reads far better than i.
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

```python
# # break exits the loop entirely -- "cherry" and everything after it is never reached.
fruits = ["apple", "banana", "cherry", "date"]

for fruit in fruits:
    if fruit == "cherry":
        break
    print(fruit)




```

```python
# continue skips just the current item and carries on with the next one.
for fruit in fruits:
    if fruit == "banana":
        continue
    print(fruit)
```

# If Else Logic

![](https://www.learnbyexample.org/wp-content/uploads/python/Python-elif-Statement-Syntax.png)


```python
# A plain if runs its body only when the condition is True, and does nothing otherwise.
number = 3

if number > 5:
    print("The number is greater than 5.")
print("hello")
```

```python
# else gives the alternative path, so exactly one of the two branches always runs.
age = 16

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
# elif adds more branches. Python checks them top to bottom and stops at the first True one.
temperature = 75

if temperature > 85:
    print("It's hot outside.")
elif temperature > 65:
    print("The weather is nice.")
else:
    print("It's a bit chilly.")

# What happens if we add the following?
# if temperature > 100:
#     print("hi!")
# else:
#     print("It's a bit chilly.")
```

```python
# and requires both conditions to be True; or requires only one of them.
x = 10
y = 20

if x > 5 and y > 15:
    print("Both conditions are met.")

if x < 5 or y > 15:
    print("At least one condition is met.")
```

```python
# in checks membership -- it works on lists, strings and dictionary keys.
fruits = ["apple", "banana", "cherry"]

if "banana" in fruits:
    print("Banana is in the list.")
else:
    print("Banana is not in the list.")
```

```python
# Each country is an object/entity, with a name and a drinking age
drinking_age = [
    {
        "name": "USA",
        "age": 21
    },
    {
        "name": "EU",
        "age": 18
    },
    {
        "name": "Japan",
        "age": 16
    },
]

# The list of people (their ages)
ages = [18, 22, 75, 3]

####################




# for each person
for person_age in ages:
  # for each country
  for country_age in drinking_age:
    # check if the person can drink in that country
    if person_age < country_age.get("age"):
      print("age = " + str(person_age) + " No drinks in " + country_age["name"])
    else:
      print("age = " + str(person_age) + " Serve a drink in " + country_age["name"])
```

# Functions

![](https://www.learnbyexample.org/wp-content/uploads/python/Python-Function-Syntax.png)

- Functions are powerful because they let us repeat the same task over and over.
- We can provide different arguments as inputs for our code to evaluate

> Why cover functions now?  We can use functions to apply (powerful) transformations to the rows across a column in our dataset.


```python
# def names the function; the indented body is what it does. Nothing runs until you call it.
def greet():
    print("Hello, world!")

greet()
```

```python
# A parameter lets the same function handle different inputs.
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
greet_person(name="Brock")   # naming the argument makes the call self-documenting
```

```python
# return hands a value back to you, so it can be stored and reused. print() only shows it on screen.
def add_numbers(a, b):
    return a + b

total = add_numbers(2, 3)
print(total)
```

# Looking Ahead

We are now going to take this fundamentals and move into data analysis with Pandas, which as we will see next week, builds on above!


```python
# pandas and matplotlib both ship with Colab, so they only need importing.
import pandas as pd
import matplotlib.pyplot as plt

# Every data file this notebook uses is read straight from the course repository.
# There is nothing to upload: DATA is the folder those files live in.
DATA = "https://raw.githubusercontent.com/elhamod/IS834_Fall_2026/main/data/"
```

```python
# Read the diamonds file straight from the course repository -- nothing to upload.
diamonds = pd.read_csv(DATA + "diamonds.csv")
diamonds.head()
```

```python
# A function that works on one row of the dataset. row['price'] reads a column, exactly like a dictionary key.
def calculate_price_per_carat(row):
    return row['price'] / row['carat']
```

```python
# .apply(axis=1) runs that function once per row and returns the results as a new column.
diamonds['price_per_carat'] = diamonds.apply(calculate_price_per_carat, axis=1)
diamonds[['carat', 'price', 'price_per_carat']].head()
```

```python
# Plot the new column against carat weight to see whether bigger stones cost more per carat.
plt.figure(figsize=(10, 6))
plt.scatter(diamonds['carat'], diamonds['price_per_carat'], alpha=0.3, c='blue')
plt.title('Price per Carat vs. Carat Weight of Diamonds')
plt.xlabel('Carat')
plt.ylabel('Price per Carat')
plt.grid(True)
plt.show()
```
